# Sales by Category Analysis

```python
sales_by_category = df.groupby("Category")["Sales"].sum()
sales_by_category.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
#  figsize=(8,5) → Makes the chart larger and easier to read.
# Better chart title.
# plt.tight_layout() → Prevents labels from getting cut off.
sales_by_category = df.groupby("Category")["Sales"].sum()

ax = sales_by_category.plot(kind="bar", figsize=(8,5))

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
sales_by_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

ax = sales_by_category.plot(kind="bar", figsize=(8,5))

plt.title("Total Sales by Category", fontsize=16)
plt.xlabel("Category", fontsize=12)
plt.ylabel("Sales", fontsize=12)

# Add value labels on top of bars
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.xticks(rotation=0)

plt.tight_layout()

plt.savefig("../visuals/sales_by_category_professional.png", dpi=300)

plt.show()
```

## Monthly Sales Trend

```python
df["Order.Date"].head()
```

```python
df["Order.Date"] = pd.to_datetime(df["Order.Date"])
```

```python
df["Month"] = df["Order.Date"].dt.to_period("M")
```

```python
df["Month"].head()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()
```

```python
monthly_sales.head()
```

```python
ax = monthly_sales.plot(kind="line", figsize=(12,5))
plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales", fontsize=12)
plt.tight_layout()
plt.savefig("../visuals/monthly_sales_trend.png", dpi=300)

plt.show()
```
